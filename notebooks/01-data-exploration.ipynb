# Data Loading 

```python
import pandas as pd

# Define paths based on your actual file list
main_file = '../data/raw/ethiopia_fi_unified_data.xlsx'
ref_file = '../data/raw/reference_codes.xlsx'

# 1. Load the main unified data
df_main = pd.read_excel(main_file, sheet_name='ethiopia_fi_unified_data')

# 2. Load the impact links (the relationships)
df_impact = pd.read_excel(main_file, sheet_name='Impact_sheet')

# 3. Load the reference codes (Excel version)
df_ref = pd.read_excel(ref_file)

print("✅ All datasets loaded successfully!")
print(f"Main Data Rows: {len(df_main)}")
print(f"Impact Links Rows: {len(df_impact)}")
print(f"Reference Codes: {len(df_ref)}")
```

# Data Exploration 

```python
# A. Count records by record_type (Should show: observation, event, target)
print("Record Type Breakdown:")
print(df_main['record_type'].value_counts())

# B. Identify the temporal range (How far back and forward does the data go?)
df_main['observation_date'] = pd.to_datetime(df_main['observation_date'])
print(f"\nDate Range: {df_main['observation_date'].min()} to {df_main['observation_date'].max()}")

# C. List unique indicators
print("\nUnique Indicator Codes:")
print(df_main['indicator_code'].unique())
```

# Data Enrichment 

```python
# Create a new observation row for 2023 Mobile Money usage
new_data = {
    "record_id": "REC_ENRICH_01",
    "pillar": "usage",
    "indicator": "Total Digital Payment Value (ETB)",
    "indicator_code": "DIGITAL_PAY_VAL",
    "value_numeric": 1100000000000, # 1.1 Trillion ETB
    "observation_date": pd.to_datetime("2023-06-30"),
    "record_type": "observation",
    "source_name": "NBE Annual Report 2022/23",
    "confidence": "high",
    "notes": "Shows massive growth in digital payments before 2024 survey"
}

# Add it to the dataframe
df_main = pd.concat([df_main, pd.DataFrame([new_data])], ignore_index=True)
print("✅ Enrichment: Added 2023 Digital Payment value.")
```

# save for the next task 

```python
# Create the processed directory if it doesn't exist
import os
os.makedirs('../data/processed', exist_ok=True)

# Save the enriched data
with pd.ExcelWriter('../data/processed/ethiopia_fi_enriched.xlsx') as writer:
    df_main.to_excel(writer, sheet_name='ethiopia_fi_unified_data', index=False)
    df_impact.to_excel(writer, sheet_name='Impact_sheet', index=False)

print("✅ Task 1 complete. Enriched file saved to data/processed/")
```
